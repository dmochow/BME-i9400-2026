**Dataset Choice and Analysis Plan**
I'm using the Diabetes 130-US hospitals dataset (Strack et al., 2014; UCI Machine Learning Repository, CC BY 4.0), which contains about 100,000 hospital encounters of patients with diabetes from 1999 to 2008. Each row is one encounter, and the label is 1 if the patient was readmitted within 30 days of discharge, which happens in about 11% of cases. The starter notebook already splits the data by patient, so every encounter of a given patient lands in the same set, and I will keep that split. Inside the training set I will use grouped cross-validation by patient for the same reason. I also plan to handle two data problems. Encounters that ended in death or hospice discharge cannot be followed by a readmission, so I will drop them and report how many rows that removes. Missing values are coded as ?, so I will convert them to proper nulls and drop weight, which is missing for almost every encounter, and I will compare a "missing" category against imputation for the other fields.
My primary metric is AUROC on the held-out test set. It measures how well a model ranks readmitted patients above the rest without depending on a threshold, and it is the usual metric for this dataset, so my results can be compared with prior work. To get confidence intervals,I'll bootstrap patients instead of encounters. That means I resample the test set many times by randomly picking whole patients (with all their encounters) and compute the AUROC each time. The spread of those scores gives me a confidence interval. I do it this way because one patient's encounters are related, so resampling single encounters would make my results look more certain than they really are.
I will try two models beyond the baseline. The first is gradient boosting which builds an ensemble of shallow decision trees where each tree corrects the errors of the previous ones. The second is a random forest, which averages many independently trained decision trees and gives a more stable, lower-tuning tree-based comparison. Both can capture nonlinear effects and interactions that logistic regression misses. I will tune both with grouped cross-validation on the training set only.
